## Q1. Sensor Readings

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a) Add 2°C to every reading using vectorization
corrected_temperature = temperature + 2
print("a) Corrected temperature readings:", corrected_temperature)

# b) Convert Celsius to Fahrenheit
fahrenheit = (9 / 5) * corrected_temperature + 32
print("b) Temperature in Fahrenheit:", fahrenheit)

# c) Identify readings greater than 32°C using Boolean indexing
greater_than_32 = corrected_temperature[corrected_temperature > 32]
print("c) Readings greater than 32°C:", greater_than_32)

# d) Count how many readings exceed 32°C
count = np.sum(corrected_temperature > 32)
print("d) Number of readings exceeding 32°C:", count)

# e) Explanation
print("\ne) Explanation:")
print("Vectorization performs operations on the complete NumPy array at once,")
print("without explicitly writing a for loop. Boolean indexing directly selects")
print("elements satisfying a condition. These operations are concise and generally")
print("more efficient than explicitly iterating through every value.")
```

## Q2. Daily Steps

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a) Total steps recorded
print("a) Total steps:", np.sum(steps))

# b) Mean number of steps
print("b) Mean steps:", np.mean(steps))

# c) Maximum and minimum values
print("c) Maximum steps:", np.max(steps))
print("   Minimum steps:", np.min(steps))

# d) Total steps for each day using axis=0
print("d) Total steps for each day:", np.sum(steps, axis=0))

# e) Total steps for each user using axis=1
print("e) Total steps for each user:", np.sum(steps, axis=1))

# f) User/day position containing maximum number of steps
max_index = np.argmax(steps)
max_position = np.unravel_index(max_index, steps.shape)

print("f) Maximum value:", steps[max_position])
print("   Position (user index, day index):", max_position)
print("   Position (1-based): User", max_position[0] + 1,
      "Day", max_position[1] + 1)
```

## Q3. NumPy Slicing, Copying, Reshaping, Flattening and Array Properties

```python
import numpy as np

# a) Create the original array
original = np.array([1, 2, 3, 4, 5, 6])
print("a) Original:", original)

# b) Slice elements from index 1 to 4
subset = original[1:5]
print("\nb) Subset:", subset)

# c) Modify the first element of subset
subset[0] = 999
print("\nc) After modifying subset:")
print("Original:", original)
print("Subset:", subset)

# d) Create another slice using .copy()
copied_array = original[1:5].copy()
copied_array[0] = 500

print("\nd) After modifying copied array:")
print("Original:", original)
print("Copied array:", copied_array)

# e) Create numbers 1 to 12 and reshape into 3 x 4
matrix = np.arange(1, 13).reshape(3, 4)
print("\ne) 3 x 4 matrix:")
print(matrix)

# f) Indexing and slicing
print("\nf) First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

# g) Flatten using flatten() and ravel()
flattened = matrix.flatten()
raveled = matrix.ravel()

print("\ng) Using flatten():", flattened)
print("Using ravel():", raveled)

# h) Modify an element returned by ravel()
raveled[0] = 1000

print("\nh) After modifying ravel():")
print("Original matrix:")
print(matrix)
print("Raveled array:")
print(raveled)

# i) Modify an element returned by flatten()
flattened[1] = 2000

print("\ni) After modifying flatten():")
print("Original matrix:")
print(matrix)
print("Flattened array:")
print(flattened)

print("\nObservation:")
print("ravel() usually returns a view, so modifying it can affect the original array.")
print("flatten() returns a copy, so modifying it does not affect the original array.")

# j) Display array properties
print("\nj) Shape:", matrix.shape)
print("ndim:", matrix.ndim)
print("size:", matrix.size)
print("dtype:", matrix.dtype)
```

## Q4. Cognitive-Assistive System and Ordinary Least Squares

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a) Display shape and dimensions of X
print("a) Shape of X:", X.shape)
print("   Dimensions of X:", X.ndim)

# b) Transpose X
XT = X.T
print("\nb) X.T:")
print(XT)
print("Explanation: Each row of X represents one user and each column")
print("represents a feature. X.T changes rows into columns and columns into rows.")

# c) Matrix product X.T @ X
XTX = X.T @ X
print("\nc) X.T @ X:")
print(XTX)

# d) Matrix inverse
XTX_inverse = np.linalg.inv(XTX)
print("\nd) Inverse of X.T @ X:")
print(XTX_inverse)

# e) Ordinary Least Squares:
# beta = (X.T @ X)^(-1) @ X.T @ y
beta = XTX_inverse @ X.T @ y

print("\ne) OLS coefficients (beta):")
print(beta)

# f) Explain coefficients
print("\nf) Coefficient interpretation:")
print("Coefficient 1 corresponds to Sleep hours.")
print("Coefficient 2 corresponds to Activity level.")
print("Coefficient 3 corresponds to Stress level.")
print("Each coefficient represents the estimated change in assistance score")
print("associated with a one-unit change in that feature, according to this OLS model.")

# g) Predict assistance score for a new user
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("\ng) New user:", new_user)
print("Predicted assistance score:", predicted_score)
```
